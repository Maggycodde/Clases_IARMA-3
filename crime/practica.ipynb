### Limpieza de datos

```python
df.isnull().sum()
```

```python
(df.isnull().sum() / len(df)) * 100
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()

print("Duplicados eliminados")
```

```python
columnas_numericas = df.select_dtypes(include=np.number).columns

columnas_numericas
```

```python
for columna in columnas_numericas:
    df[columna] = df[columna].fillna(df[columna].mean())

print("Valores numéricos corregidos")
```

```python
columnas_categoricas = df.select_dtypes(include='object').columns

columnas_categoricas
```

```python
for columna in columnas_categoricas:
    moda = df[columna].mode()[0]
    df[columna] = df[columna].fillna(moda)

print("Valores categóricos corregidos")
```

### Transformacion de variables

### Creación de Grupo_Temporada (Invierno, Primavera, Verano, Otoño)

```python
df["Grupo_Temporada"] = pd.cut(
    df["MONTH"],
    bins=[0,3,6,9,12],
    labels=[
        "Invierno",
        "Primavera",
        "Verano",
        "Otoño"
    ]
)

df[["MONTH","Grupo_Temporada"]].head()
```

### Creación de Grupo_Horario (Madrugada, Mañana, Tarde, Noche)

```python
df["Grupo_Horario"] = pd.cut(
    df["HOUR"],
    bins=[0,6,12,18,24],
    labels=[
        "Madrugada",
        "Mañana",
        "Tarde",
        "Noche"
    ]
)

df[["HOUR","Grupo_Horario"]].head()
```

### Deteccion de outliers

```python
df["YEAR"].mean()
```

```python
df["HOUR"].mean()
```

```python
df.groupby("HOUR")["YEAR"].mean()
```

```python
df.groupby("INCIDENT_NUMBER")["YEAR"].mean()
```

```python
df.groupby("STREET")["YEAR"].mean()
```

```python
Q1 = df["HOUR"].quantile(0.25)
Q3 = df["HOUR"].quantile(0.75)

IQR = Q3 - Q1

limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

print("Límite inferior:", limite_inferior)
print("Límite superior:", limite_superior)
```

```python
outliers = df[
    (df["HOUR"] < limite_inferior) |
    (df["HOUR"] > limite_superior)
]

outliers
```

```python
print("Outliers encontrados:", len(outliers))
```

### Distribucion espacial de crimenes por coordenadas Long/Lat (Muestra de 8000)

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("crime.csv", encoding="latin-1", low_memory=False)

df_mapa = df[(df["Lat"] > 0) & (df["Long"] < 0)].copy()
print(f"Registros válidos: {len(df_mapa)}")

muestra = df_mapa.sample(n=8000, random_state=42)
print(f"Registros en la muestra: {len(muestra)}")

fig = px.scatter_map(
    muestra,
    lat="Lat",
    lon="Long",
    color="UCR_PART",
    hover_data={
        "DISTRICT": True,
        "OFFENSE_CODE_GROUP": True,
        "STREET": True,
        "YEAR": True,
        "Lat": False,
        "Long": False
    },
    zoom=11.5,
    height=700,
    title=f"Distribución geográfica de crímenes en Boston (n={len(muestra)})",
    opacity=0.5,
    color_discrete_map={
        "Part One": "#e74c3c",
        "Part Two": "#f39c12",
        "Part Three": "#3498db"
    }
)


fig.update_layout(
    map_style="carto-darkmatter",   
    map_center={"lat": 42.3601, "lon": -71.0589},  
    margin={"r": 0, "t": 50, "l": 0, "b": 0},
    legend_title_text="Gravedad del delito (UCR)"
)

fig.update_traces(marker=dict(size=5))
fig.show()
```

### Distribucion de la categoria de delito (UCR_PART) por Distrito

```python
tabla = pd.crosstab(df['DISTRICT'], df['UCR_PART'])

tabla.plot(kind='bar', stacked=True, figsize=(10,5), colormap='Set2')
plt.title('Distribución de UCR_PART por Distrito')
plt.ylabel('N° de crímenes')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show() 
```
